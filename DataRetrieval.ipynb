```python
import requests
import time

url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_day.geojson"

seen_ids = set()

while True:
    response = requests.get(url)
    data = response.json()

    new_events = []

    for feature in data["features"]:
        earthquake_id = feature["id"]

        if earthquake_id not in seen_ids:
            earthquake = {
                "id": earthquake_id,
                "magnitude": feature["properties"]["mag"],
                "place": feature["properties"]["place"],
                "time": feature["properties"]["time"],
                "longitude": feature["geometry"]["coordinates"][0],
                "latitude": feature["geometry"]["coordinates"][1],
                "depth": feature["geometry"]["coordinates"][2]
            }

            new_events.append(earthquake)
            seen_ids.add(earthquake_id)

    print("New earthquakes:", len(new_events))

    for event in new_events:
        print(event)

    time.sleep(60)
```

```python
import os

os.environ["PYSPARK_SUBMIT_ARGS"] = (
    "--packages "
    "org.apache.spark:spark-sql-kafka-0-10_2.13:4.0.0 "
    "pyspark-shell"
)

print("Spark-Kafka connector configured!")
```

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("EarthquakeStreaming")
    .master("local[*]")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")

print("Spark Session created successfully!")
```

```python
kafka_df = (
    spark.readStream
    .format("kafka")
    .option("kafka.bootstrap.servers", "localhost:9092")
    .option("subscribe", "earthquakes")
    .option("startingOffsets", "earliest")
    .load()
)

print("Kafka streaming DataFrame created!")
```

```python
from pyspark.sql.functions import col

messages = kafka_df.select(
    col("value").cast("string").alias("value")
)

print("Kafka messages prepared for processing!")
```

```python
query = (
    messages.writeStream
    .format("console")
    .outputMode("append")
    .option("truncate", False)
    .start()
)

query.awaitTermination(10)
```

```python
print("Query active:", query.isActive)
print("Query status:", query.status)
print("Query exception:", query.exception())
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    DoubleType,
    LongType
)

earthquake_schema = StructType([
    StructField("id", StringType(), True),
    StructField("magnitude", DoubleType(), True),
    StructField("place", StringType(), True),
    StructField("time", LongType(), True),
    StructField("longitude", DoubleType(), True),
    StructField("latitude", DoubleType(), True),
    StructField("depth", DoubleType(), True)
])

print("Earthquake schema created!")
```

```python
from pyspark.sql.functions import from_json, col

parsed_df = (
    kafka_df
    .select(
        from_json(
            col("value").cast("string"),
            earthquake_schema
        ).alias("data")
    )
    .select("data.*")
)

print("JSON parsed into structured columns!")
```

```python
parsed_query = (
    parsed_df.writeStream
    .format("console")
    .outputMode("append")
    .option("truncate", False)
    .start()
)

print("Structured Spark stream started.")
```

```python
import time

time.sleep(10)

print("Recent batches:")
print(parsed_query.recentProgress)
```

```python
parsed_query.stop()

print("Structured Spark stream stopped.")
```

```python
import requests
import time
from kafka import KafkaProducer
import json

USGS_URL = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/all_day.geojson"

producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda v: json.dumps(v).encode("utf-8")
)

seen_ids = set()

print("USGS → Kafka ingestion started")
```

```python
while True:
    response = requests.get(USGS_URL)
    data = response.json()

    new_count = 0

    for feature in data["features"]:
        earthquake_id = feature["id"]

        if earthquake_id not in seen_ids:
            event = {
                "id": earthquake_id,
                "magnitude": feature["properties"]["mag"],
                "place": feature["properties"]["place"],
                "time": feature["properties"]["time"],
                "longitude": feature["geometry"]["coordinates"][0],
                "latitude": feature["geometry"]["coordinates"][1],
                "depth": feature["geometry"]["coordinates"][2]
            }

            producer.send("earthquakes", value=event)
            seen_ids.add(earthquake_id)
            new_count += 1

            print("Sent to Kafka:", event)

    producer.flush()

    print("New events sent:", new_count)
    print("-" * 50)

    time.sleep(60)
```

```python
parsed_query = (
    parsed_df.writeStream
    .format("console")
    .outputMode("append")
    .option("truncate", False)
    .start()
)

print("Spark streaming is running...")
```
